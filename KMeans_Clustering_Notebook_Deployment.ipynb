```python
## 20. Loading the Saved Model
import pickle
import pandas as pd

filename = "KMeans_Clustering_Model.sav"
load_model = pickle.load(open(filename, 'rb'))
print("Model loaded successfully.")
```

```python
## 21. Taking New Customer Input
annual_income = float(input("Annual Income: "))
spending_score = float(input("Spending Score: "))
new_customer = [[annual_income, spending_score]]
```

```python
## 22. Generating the Cluster Prediction
prediction = load_model.predict(new_customer)
print("Predicted Cluster:", prediction[0])
```
