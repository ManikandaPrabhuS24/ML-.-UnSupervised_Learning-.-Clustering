```python
## 1. Loading the Saved Model
import pickle
import numpy as np
load_model = pickle.load(open("BIRCH_Clustering_Model.sav","rb"))
print("Model loaded successfully.")
```

```python
## 2. Preparing New Customer Input
annual_income = float(input("Annual Income (k$): "))
spending_score = float(input("Spending Score (1-100): "))
new_customer = [[annual_income, spending_score]]
```

```python
## 4. Generating the Cluster Prediction
prediction = load_model.predict(new_customer)
print("Predicted Cluster:", prediction[0])
```
